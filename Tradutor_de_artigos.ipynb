```python
from azure.core.credentials import AzureKeyCredential
from azure.ai.translation.text import TextTranslationClient

chave = "SUA_CHAVE_AQUI"
ponto_de_extremidade = "SEU_PONTO_DE_EXTREMIDADE_AQUI"
regiao = "SUA_REGIAO_AQUI"

idioma_de_destino = "pt"

artigo_original = """
PASTE YOUR TECHNICAL ARTICLE TEXT HERE.
For example:
The Transformer architecture is based on a self-attention mechanism
that allows the model to weigh the importance of different words in
the input sequence. This has proven to be highly effective for a
wide range of natural language processing tasks, including machine
translation and text summarization. Unlike recurrent neural networks (RNNs),
Transformers do not process data in order, which allows for
significant parallelization and reduced training times.
"""
```

```python
credencial = AzureKeyCredential(chave)
cliente_traducao = TextTranslationClient(endpoint=ponto_de_extremidade, credential=credencial, region=regiao)

resposta = cliente_traducao.translate(content=[artigo_original], to=[idioma_de_destino])
traducao = resposta[0].translations[0]

print("--- Artigo Traduzido ---")
print(f"Idioma Detectado: {resposta[0].detected_language.language}")
print(f"Pontuação de Confiança: {resposta[0].detected_language.score}\n")
print(traducao.text)
```
